1. What is the purpose of a PivotTable in Excel?

   - A PivotTable is used to summarize, analyze, and reorganize large datasets quickly.

   - It helps in claculating total, averages, comparisons, and trends without changing the original data.

2. Create a PivotTable to show Total Sales by Region

   - Steps:

   1. Select the dataset

   2. Go to Insert -> PivotTable

   3. Drag Region to Rows

   4. Drag Sales to Values (set as Sum)

   - Result:

   - The PivotTable shows Total Sales for each Region (East, West, North, South).

   

3. Apply Conditional Formatting to highlight sales greater than 400

   - Steps:

   1. Select the Sales column

   2. Go to Home -> Conditional Formatting

   3. Choose Highlight Cells Rules -> Greater Than

   4. Enter 400 and apply a color

   - Result:

   - All sales values greater than 400 are highlighted.

4. Difference between Sorting and Filtering (with example)

   - Sorting arranges data in ascending or descending order.

   - Example: Sorting sales from highest to lowest.

   - Filtering shows only selected data based on a condition.

   - Example: Showing only records where Region = North.

   

5. Create a PivotChart showing Montly Sales Trend

   - Steps:

   1. Create a PivotTable

   2. Drag Data to Rows and group by Months

   3. Drag Sales to Values

   4. Insert -> PivotChart -> Line Chart

   - Result:

   - The PivotChart displays monthly sales trend from Jan to Jun 2024.

   

6. Use a Slicar to filter data by Product

   - Steps:

   1. Select PivotTable

   2. Go to Insert -> Slicer

   3. Choose Product

   4. Select Apples, Oranges, or Mangoes

   - Observation:

   - Sales values change dynamically based on the selected product.

7. Why are Dashboards important for decision-making in business analysis?

   - Dashboards provide a visual summary of key metrics, making data easy to understand.

   - They help management track performace, indentify trends, and take quick decisions.

8. Create a Dashboard that includes

- Sales by Region (PivotTable + Chart)

- Product Filter (Slicer)

- Monthly Trend (Line Chart)

   - Explanation:

   - The dashboards combines multiple visuals to provide a complete overview of sales performance in one screen.

9. Groups dates into Months and calculate average sales per month

   - Steps:

   1. Right-click Date in PivotTable -> Group

   2. Select Months

   3. Change Sales value setting to Average

   - Result:

   - The PivotTable shows average monthly sales.

10. Combine Apples and Oranges into one group called "Fruits"

    - Steps:

    1. Select Apples and Oranges

    2. Right-click -> Oranges

    3. Rename the group as Fruits

    - Result:

    - Sales of Fruits can be compared with other products like Mangoes.
